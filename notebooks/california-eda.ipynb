```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
import pandas as pd

# 데이터 로드 및 DataFrame 변환
california = fetch_california_housing(as_frame=True)
df = california.frame

# 시각화 설정
plt.figure(figsize=(12, 5))

# 1. 히스토그램 (값의 분포와 5.0 캡 확인)
plt.subplot(1, 2, 1)
sns.histplot(data=df, x='MedHouseVal', bins=50, kde=True, color='royalblue')
plt.title('Distribution of MedHouseVal (with Ceiling Cap)')
plt.xlabel('Median House Value ($100,000s)')
plt.ylabel('Count')

# 2. 박스플롯 (이상치 및 사분위수 확인)
plt.subplot(1, 2, 2)
sns.boxplot(data=df, y='MedHouseVal', color='lightcoral')
plt.title('Boxplot of MedHouseVal')
plt.ylabel('Median House Value ($100,000s)')

plt.tight_layout()
plt.show()
```

```python
df_cleaned = df[~(df["MedHouseVal"] > 5.0)]
df_cleaned.to_csv("../data/fetch_california_housing_clean.csv", index = False)
```
